```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor 
import joblib

df_bank = pd.read_csv('bank_transactions_data_2_augmented_clean_2.csv')


features = ['CustomerAge', 'TransactionAmount', 'LoginAttempts', 'TransactionDuration']
X = df_bank[features]
y = df_bank['AccountBalance'] 

modelo_banco = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_banco.fit(X, y)

joblib.dump(modelo_banco, 'modelo_bancario.pkl')
print("Modelo Bancário salvo com sucesso!")
```
