```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor # Ou Classifier se tiver a coluna de fraude
import joblib

# 1. Carregar o dataset bancário
df_bank = pd.read_csv('bank_transactions_data_2_augmented_clean_2.csv')

# 2. Selecionar colunas que influenciam o valor (Exemplo de Regressão de Saldo)
# Vamos usar Idade, Valor da Transação e Tentativas de Login
features = ['CustomerAge', 'TransactionAmount', 'LoginAttempts', 'TransactionDuration']
X = df_bank[features]
y = df_bank['AccountBalance'] # Prevendo o saldo final baseado na transação

# 3. Treinar o modelo
modelo_banco = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_banco.fit(X, y)

# 4. Salvar
joblib.dump(modelo_banco, 'modelo_bancario.pkl')
print("✅ Modelo Bancário salvo com sucesso!")
```
