```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import joblib

df = pd.read_csv('train.csv')

# Selecionar colunas e tratar dados
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Fare']
X = df[features].copy()
y = df['Survived']

# Tratando os dados (Exatamente como o modelo espera)
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1})
X['Age'] = X['Age'].fillna(X['Age'].median())
X['Fare'] = X['Fare'].fillna(X['Fare'].median())

print("Dados preparados!")
X.head() # Para você visualizar no Jupyter
```

```python
# Criar e treinar o modelo
modelo_titanic = RandomForestClassifier(n_estimators=100, random_state=42)
modelo_titanic.fit(X, y)

# SALVAR O MODELO (Isso cria o arquivo que o Streamlit vai usar)
joblib.dump(modelo_titanic, 'modelo_titanic.pkl')
print("Arquivo 'modelo_titanic.pkl' gerado com sucesso!")
```
